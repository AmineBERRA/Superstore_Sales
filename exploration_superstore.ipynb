Nettoyage Dataset

```python
import pandas as pd

# path to the local CSV file
file_path = r"D:\WORK\Projets_Portfolio\Superstore Sales Portfolio\Superstore_Sales\Superstore_Sales.csv"

# read it directly with pandas, specifying encoding to avoid UnicodeDecodeError
df = pd.read_csv(file_path, encoding='latin1')  # or 'iso-8859-1', depending on file

print("First 5 records:", df.head())
```

```python
df.info()
print("Missing values per column:\n", df.isnull().sum())
```

```python
# Check for duplicates
duplicates = df.duplicated().sum()
print(f"Number of duplicate records: {duplicates}")
```

```python
# Drop columns that are not needed for analysis 
columns_to_drop = ['Postal Code']  # Trop de valeurs manquantes pour une colonne pas essentielle
df.drop(columns=columns_to_drop, inplace=True)
print("Columns after dropping:", df.columns)
```

```python
df_datetime = df
# Convert 'Order Date' and 'Ship Date' to datetime
df_datetime['Order Date'] = pd.to_datetime(df_datetime['Order Date'], dayfirst=True,format= 'mixed')  # dayfirst=True for 'dd/mm/yyyy' format
df_datetime['Ship Date'] = pd.to_datetime(df_datetime['Ship Date'], dayfirst=True,format= 'mixed')
print("Data types after conversion:\n", df_datetime.dtypes) 
print("First 5 records after date conversion:\n", df_datetime[['Order Date', 'Ship Date']].head())
#check for any missing values in the date columns after conversion
print("Missing values in 'Order Date':", df_datetime['Order Date'].isnull().sum())
print("Missing values in 'Ship Date':", df_datetime['Ship Date'].isnull().sum())
```

```python
#df_datetime = df # Use df_datetime for further analysis since it has the date columns converted to datetime format
df = df_datetime
df.head()
df.isnull().sum()
```

Feature Engineering & Statistiques descriptives

```python
#df['Sales'] & df['Profit'] statistics
print("Sales column statistics:\n", df['Sales'].describe().round(2))
print("\nProfit column statistics:\n", df['Profit'].describe().round(2))
```

```python
#calculer la marge 
marge_percentage = df['Profit'] / df['Sales'] * 100
print("First 5 records of Marge Percentage:\n", marge_percentage.head().round(2), "%")

df['Marge'] = marge_percentage
print("First 5 records with Marge:\n", df[['Sales', 'Marge']].head().round(2))
```

```python
#Separate 'Order Date' into 'Year', 'Month'columns for more granular time-based analysis
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month 
print("First 5 records with Year and Month:\n", df[['Order Date', 'Year', 'Month']].head())
```

```python
#make marge percentage more readable by rounding to 2 decimal places
df['Marge'] = df['Marge'].round(2)
df.head()
```

```python
df.to_csv(r"D:\WORK\Projets_Portfolio\Superstore Sales Portfolio\Superstore_Sales\Superstore_Sales_Cleaned.csv", index=False)
```
